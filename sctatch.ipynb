```python
import os 
import yaml 
import numpy as np 
from pathlib import Path
from data_manager.arxiv_items import Item
```

```python
with open (Path("clix_pages") / "resources" / "available_arxiv_categories.yaml") as f:
    data = yaml.load(f, Loader=yaml.FullLoader)
```

```python
item_list = []
count = 0
for key, value in data.items():
    item_list.append(
        Item(id_=count, title_=key, value_=value)
    )
    count += 1
```

```python
for item in item_list:
    item.print_item(ntab=0)
```

```python
for item in item_list:
    print(item.get_item()["title"])
    for next_item in item.get_item()['value']:
        print(next_item.get_item())
```
